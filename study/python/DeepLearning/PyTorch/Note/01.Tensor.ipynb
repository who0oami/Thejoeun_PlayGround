#### Tensor
- 머신러닝과 딥러닝에서 사용되는 기본적인 구조
- 다차원 배열과 숫자데이터를 저장하고 조작하는데 사용

##### Tensor의 특징
- 다차원 배열 : Tensor는 1차원(벡터), 2차원(행렬), 3차원 이상의 다차원 데이터를 표현
- GPU 지원 : Tensor는 GPU에서 연산을 수행할 수 있어 빠른 계산이 가능
- 자동 미분 :  Tensor는 자동 미분(AutoGrad) 기능을 지원하여 딥러닝 모델의 학습에 필요한 기울기 계산을 자동으로 수행 한다.

```python
# Tensor의 차원변경 : squeeze, unsqueeze

t = torch.tensor([1,2,3])
print(t)
print(t.shape)
print("-" * 15)

t_unsqueezed = t.unsqueeze(0) # 지정한 위치에 크기가 1인 차원을 추가 
print(t_unsqueezed)
print(t_unsqueezed.shape)
print("-" * 15)

t_unsqueezed_squeeze = t_unsqueezed.squeeze(0) # 지정한 위치에 크기가 1인 차원을 제거(그 차원의 크기가 1일때만 제거)
print(t_unsqueezed)
print(t_unsqueezed.shape)
print("-" * 15)
```

#### DataSet
- 데이터의 전처리와 모델학습을 효율적으로 수행하기 위한 핵심 구성 요소
- DataSet 데이터와 라벨을 저장
- DataLoader는 이러한 데이터를 쉽게 순회 할 수 있게 한다. (순회 : 섞어두고서 쓰는 거-> 과대적합이 덜 일어나게 하기 위해서)

#### DataSet과 DataLoader의 개념

##### DataSet
- 역할 : 데이터 샘플과 그에 해당하는 레이블을 저장
- 주요 Method
    - init : 데이터셋 초기화 및 데이터 로드 
    - len : 데이터셋의 샘플 갯수를 반환
    - getitem : 주어진 인덱스에 해당하는 샘플을 반환

##### DataLoader
- 역할 : DataSet을 Batch 단위로 쉽게 접근할 수 있도록 하는 iterable  :-> iterable 객체를 제공 (리스트)데이터 하나에 여러개 들어있는 데이터
- 주요 Method
    batch_size : 한번에 불러올 데이터의 크기
    - shuffle : 데이터를 무작위로 섞을지 여부
    - num_workers : 데이터를 로드할때 사용한 서브 프로세스의 수 

```python
# 클래스 만들기
# 기본 데이터셋 클래스 정의
from torch.utils.data import Dataset # Dataset이라는 기본 클래스를 상속 받음

class SimpleDataset(Dataset):
    def __init__(self, data, labels): # 시스템에서 쓰는 건 다 언더바 두개씀/ 그래서 우리가 만드는거에 언더바 쓰면 안된다 -> 한개까지는 가능
        self.data = data
        self.labels = labels

    def __len__(self):
        return len(self.data)
    
    def __getitem__(self, idx):
        return self.data[idx], self.labels[idx], 
```

```python
data = [[1,2], [3,4], [5,6]]
labels = [0,1,0]
dataset = SimpleDataset(data, labels) 
```

```python
print(dataset.data)
print(dataset.labels)
print(len(dataset))
print(dataset[1])
```

```python
# 데이터셋은 리스트도 아니고 객체이다
```

```python
# 기본 Dataloader 사용 : Shuffle 하지 않고 전체를 한번에 사용하자 batch_size=3

from torch.utils.data import DataLoader

dataloder = DataLoader(dataset, batch_size=3, shuffle=False)

for batch in dataloder:
    print(batch)
```

```python
# shuffle하고 2개씩 불러오기
dataloder = DataLoader(dataset, batch_size=2, shuffle=True)

for batch in dataloder:
    print(batch)

# tendor가 두개씩임
```
