#### 패션 MNIST
- class : 10 개
- 28 * 28 pixel
- sample : 70,000 개


```python
from tensorflow import keras

(train_input, train_target), (test_input, test_target) = keras.datasets.fashion_mnist.load_data()
```

```python
# train -> train, vaild 분리
from sklearn.model_selection import train_test_split

train_scaled = train_input / 255.0 # 0 ~ 255 -> 0 ~ 1 사이로 스케일링 / 255.0인 이유 : 255는 8비트로 표현할 수 있는 최대값이기 때문
train_scaled = train_scaled.reshape(-1, 28 * 28) # 28 * 28 = 784 # 3차원 -> 2차원

train_scaled, val_scaled, train_target, val_target = train_test_split(
                                                        train_scaled,
                                                        train_target, 
                                                        test_size=0.2, 
                                                        random_state=42
)
```

- 입력층 : 배열의 갯수가 28 * 28 = 784 개의 입력층이 필요하다 (노드가 784 개)
- 은닉층 : 갯수는 정해져 있지 않으나 , 출력층 보다 많아야 한다 (2의 배수), Hper Parameter
- 출력층 : 분류인 경우, 다중분류는 Softmax, 이진분류는 Sigmoid

- 훈련시 optimizer를 사용하여 최적화 알고리즘을 구성한다.   
- 보통 사용하는 것이 확률적 경사 하강법 optimizer이다.  
-- SGD(learning_rate=0.01) : 기본 경사 하강법 옵티마이저    
-- 적응적 학습률 옵티마이져 : RMAprop, Adam <- (Adagrad): 아담 전에 사용하던 옵티마이저, 학습률이 점점 감소하는 단점이 있다.


```python
# 적용방법 (확률적 경사 하강법) - 1

model.compile(
    optimizer = 'sgd', # 확률적 경사 하강법 옵티마이저_# 이렇게 안적으면 기본적응로 경사 하강법임
    loss="sparse_categorical_crossentropy", # 손실 함수는 sparse_categorical_crossentropy (다중 클래스 분류 문제에서 사용)
    metrics=["accuracy"] # 평가 지표는 정확도
)
```

```python
# 적용방법 (확률적 경사 하강법) - 2

sgd = keras.optimizers.SGD() # 확률적 경사 하강법 옵티마이저, learning_rate는 학습률을 의미한다.
model.compile(
    optimizer = sgd, # 변수 이름만 주면 되니까 -> if 문 쓸 수 있어 이렇게 사용하는 거 추천
    loss="sparse_categorical_crossentropy", # 손실 함수는 sparse_categorical_crossentropy (다중 클래스 분류 문제에서 사용)
    metrics=["accuracy"] # 평가 지표는 정확도
)
```

```python
# 적용방법 (확률적 경사 하강법) - 3

sgd = keras.optimizers.SGD(learning_rate=0.01) # learning_rate = 0.01
model.compile(
    optimizer = sgd, # 변수 이름만 주면 되니까 -> if 문 쓸 수 있어 이렇게 사용하는 거 추천
    loss="sparse_categorical_crossentropy", # 손실 함수는 sparse_categorical_crossentropy (다중 클래스 분류 문제에서 사용)
    metrics=["accuracy"] # 평가 지표는 정확도
)
```

#### 딥러닝 구성
##### 딥러닝 작업 흐름
###### 1. 입력층(Input Layer)
- 입력 데이터를 받는 층 입니다. 데이터는 숫자로 변환된 상태로 들어 옵니다.   
- 입력층은 데이터의 특징이나 변수를 신경망에 전달 합니다.   

###### 2. 은닉층(Hidden Layer)
- 입력층에서 받은 데이터를 처리하는 중간 단계 입니다. 하나 이상의 은닉층이 있을 수 있습니다.   
- 각 은닉층에서 활성화 함수(Activation Function)가 적용 됩니다.
- 이 함수는 각 노드에서 계산된 값을 비선형적으로 변환 (확률값으로 변환) 하여 다음 층으로 전달 합니다.
- 대표적인 활성화 함수는 ReLu, Sigmoid, Tanh 등이 있습니다.

###### 3. 출력층(Output Layer)
- 은닉층에서 전달된 데이터를 최종적으로 처리하여 결과를 내는 층 입니다.
- 분류 문제라면 Softmax 같은 활성화 함수가 사용되어 예측 확률을 계산 합니다.

##### 4. 손실함수(Loss Function)
- 출력층에서 나온 결과와 실제 값(정답)을 비교하여 오류를 계산합니다.   
- 손실함수는 예측이 얼마나 잘못되었는지 평가합니다.   
- 대표적인 손실함수로는 회귀 문제에 사용되는 MSE(Mean Squared Error)와 분류 문제에 사용되는 Cross_Entropy(교차 엔트로피) 등이 있습니다.

##### 5. 최적화 함수(Optimization Function)
- 손실함수에서 계산된 오류를 바탕으로 네트워크의 가중치를 업데이트 합니다. 이를 통해 모델이 더 좋은 예측을 할 수 있도록 학습이 진행 됩니다 (네트워크: 입력층, 은닉층, 출력층 노드와 연결된 가중치)
- 대표적인 최적화 함수로는 확률적 경사 하강법(Stochastic Gradient Descent : SGD)이나 Adam 등이 있습니다.
- 최적화 함수는 역전파(back propagation) 과정에서 가중치와 편향을 조정하는 역할을 합니다.

이 과정이 여러번 반복되면서 신경망은 학습을 통해 점점 더 좋은 결과를 내놓게 됩니다.
