#### 딥러닝 이란
- 딥러닝은 심층신경망(Deep Neural Network: DNN) 이라고도 불리운다.
- 뉴런(노드)으로 구성된 레이어 여러개 연결해서 구성한 네트워크
- 대표적인 예로 CNN(Convolutional Neural Network: 사물을 이해하는 과정), RNN(Recurrent Neural Network: 문맥 이해하는 과정)


#### 딥러닝의 학습
- 학습목표 :  모델에 입력값을 넣었을때 출력값이 최대한 정답과 일치하게 하는 것.
- 학습 : 최초 딥러닝 모델에 매개변수(weight, bias)를 무작위로 부여한 후, 반복학습을 통해서 모델의 출력값이 최대한 정답과 일치되게 매개변수를 조정하는 과정

##### 순전파(forward propagation)
- 왼쪽에서 오른쪽으로 흘러가는 과정
- 순전파에 의해 딥러닝의 출력값(y^:예측값)이 결정, 정답은 (y) 

##### 손실함수 (Loss Function)
- 출력값과 정답의 차이
- 출력값과 정답이 일치할 수록 손실함수의 값은 작아진다.
- 보통 회귀에는 평균제곱오차(Mean Squared Error: MSE),
- 분류 문제에는 크로스 엔트로피(Cross Entropy)를 사용.
- 매개변수(weight, Bais) 를 조절해서 손실함수의 값을 최저로 만드는 과정을 최적화(Optimization)이라 한다.
- 최적화 작업은 Optimizer를 통해 이뤄지며 Optimizer는 역전파(back propagation)과정을 수행해서 딥러닝 모델의 매개변수를 최적화 한다.

##### 최적화(Optimaization)
- 대표적인 방법은 경사하강법(batch gradient descent)
- 반복적으로 손실함수에 대한 모델 매개변수의 미분값(기울기)을 구한 후 그 미분값의 반대방향으로 매개변수를 조절해 나가면 결국에는 최저 손실함수값에 도달한다.

##### 역전파 (Back propagation)
- Optimizer는 손실함수의 값을 최소화 하기 위해 역전파를 사용해 딥러닝 모델의 모든 매개변수를 변경한다.

#### 학습률(Learning Rate)
- 학습률을 크게 설정하면 매개 변수 변경치가 커져서 정답으로 수렴하지 않을 수도 있고, 너무 작으면 학습 시간이 상당히 오래 걸릴 수 있다.
