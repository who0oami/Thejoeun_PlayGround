#### Data 정규화

```python
# weight, height
csv['weight'] = csv['weight'] / csv['weight'].max()
csv['height'] = csv['height'] / csv['height'].max()

csv.head()
```

```python
# label
bclass = {
    'thin' : [1,0,0],
    'normal' : [0,1,0],
    'fat' : [0,0,1]
}

y = []
for v in csv['label']:
    # print(v)
    y.append(bclass[v])
```

---
#### 훈련용과 테스트 

```python
# Train
X_train = csv[['height', 'weight']][:15000]
y_train = y[:15000]
```

```python
# Test Data
X_test = csv[['height', 'weight']][15001:]
y_test = y[15001:]
```

#### Deep Learning Model 만들기

```python
# 입력층 : 2개
# 은닉층 : 512, activation = 'relu'
# 출력층 : 3개 ,activation = 'softmax' : 다중 분류 문제이기 때문에 softmax 사용
# loss : optimizer  = 'rmsprop', loss = 'categorical_crossentropy' : 다중 분류 문제이기 때문에 categorical_crossentropy 사용
```

```python
from tensorflow import keras
from tensorflow.keras.layers import Input
# 입력층 : 3 
# 은닉층 : 32
# 출력층 : 1
# activation : signoid, tanh ,relu,leakyRelu


model = keras.Sequential()
model.add(Input(shape=(2,)))
model.add(keras.layers.Dense(512,activation='relu')) # 은닉층 : 2의 배수
model.add(keras.layers.Dense(3,activation='softmax'))

```

```python
model.compile(
    optimizer= 'rmsprop',
    loss ='categorical_crossentropy',
    metrics =['accuracy']
)
```

```python
# 학습 시키기 
import numpy as np
model.fit(
    np.array(X_train),
    np.array(y_train),
    epochs = 100, # 처음에 5로 해본 뒤 , 50번씩 해보기
    
)
```

```python
model.evaluate(np.array(X_train),np.array(y_train))
```

```python
model.evaluate(np.array(X_test),np.array(y_test))
```

```python
model = keras.Sequential()
model.add(Input(shape=(2,)))
model.add(keras.layers.Dense(256,activation='relu')) # 입력층보다 많아야 함
model.add(keras.layers.Dense(128,activation='leaky_relu')) # 이렇게 추가해서 쓸 수 있음 _ 예측력이 좋아졌을떄 추가 하고, 떨어지면 안하는 게 좋음 / 처음엔 반절부터 시작해야 함
model.add(keras.layers.Dense(3,activation='softmax'))
```

```python
model.compile(
    optimizer= 'rmsprop',
    loss ='categorical_crossentropy',
    metrics =['accuracy']
)
```

```python
# 학습 시키기 
import numpy as np
model.fit(
    np.array(X_train),
    np.array(y_train),
    epochs = 100, # 처음에 5로 해본 뒤 , 50번씩 해보기
    
)
```

```python
model.evaluate(np.array(X_train),np.array(y_train))
```
