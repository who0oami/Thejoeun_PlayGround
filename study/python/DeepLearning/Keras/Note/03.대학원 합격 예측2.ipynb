#### Train Data 만들기 

```python
data.head()
```

```python
x = data[['gre','gpa','rank']]
x.head()
```

```python
y= data['admit']
y.head()
```

#### Deep Learning Model 만들기 

```python
from tensorflow import keras
from tensorflow.keras.layers import Input
# 입력층 : 3 
# 은닉층 : 32
# 출력층 : 1
# activation : signoid, tanh ,relu,leakyRelu


model = keras.Sequential()
model.add(Input(shape=(3,)))
model.add(keras.layers.Dense(256,activation='tanh')) # 은닉층 : 2의 배수
model.add(keras.layers.Dense(1,activation='sigmoid'))

```

```python
model.compile(
    optimizer= 'adam',
    loss ='binary_crossentropy',
    metrics =['accuracy']
)

# optimizer : learning rate을 알맞게 조정해줌
# : adam , adagrad, adadelta ,rmsprop , sgd


# loss : 손실함수 
# : binary_crossentropy : 분류 및 확률 예측

# metrics : 모델의 평가 기준 
```

```python
# 학습 시키기 
import numpy as np
model.fit(
    np.array(x),
    np.array(y),
    epochs = 1000,
    
)
```

```python
model.evaluate(np.array(x),np.array(y))
```

#### 예측하기 

```python
# gre : 700, 학점 : 3.7 , Rank가 4인학교
model.predict(
    np.array(
        [
        [700,4.5,1],
        [700,3.7,2],
        [700,3.7,3],
        [700,3.7,4],
        ]
    )
)
```
