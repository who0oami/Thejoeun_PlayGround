#### 레벤슈타인 거리(Lvenshtein distance)
- 두 개의 문자열이 어느 정도 다른지를 나타내는 것으로 편집 거리(Edit Distance)라고도 부르다.
- 의학 분야에서는 DNA 배열의 유사성을 판단할 떄도 사용.


```python
from lvenshtein import Lvenshtein
lv = Lvenshtein() # 인스턴스 하나 만들어서 이렇게 씀
```

```python
# 함수 잘 도는 지 테스트
print(lv.calc_distance("가나다라","가마바라")) # 틀린 거리를 보여줌
```

```python
# 신촌역과 가장 근접한 순서로 정렬

samples = ['신발', '마곡역', '신천역', '신천군', '신촌역']
base = samples[4] # 기준값을 줘야 함 _ 신촌역을 기준값으로 줌

r = sorted(samples, key= lambda n: lv.calc_distance(base, n))

for n in r:
    print(lv.calc_distance(base, n), n) # 기준이랑 틀린 갯수를 알려줌

```

#### N-gram 유사도
- 이웃한 N개의 문자     
- 서로 다른 2개의 문장을 N-gram으로 비교해보면 출현하는 단어의 종류와 빈도가 확인가능   
- 논문 도용 등을 확인 

```python
from ngram import Ngram
ngram = Ngram()
```

```python
# 2문장(gram) 나누기
a = "오늘 강남에서 맛있는 스파게티를 먹었다."
b= "강남에서 먹었던 오늘의 스파게티는 맛있었다."
print(ngram.ngram(a,2))
print(ngram.ngram(b,2))
```

```python
# 3문장으로 나누기
print(ngram.ngram(a,3))
print(ngram.ngram(b,3))
```

```python
# 유사도 비교

r2 = ngram.diff_ngram(a, b, 2)
print("2gram 유사도 :", r2)

r3 = ngram.diff_ngram(a, b, 3)
print("3gram 유사도 :", r3)
```

```python
# 퀴즈
a = "머신러닝은 매우 재미있는 기술이라 공부하고 있습니다."
b = "공부하면 재미있는 기술이라 머신러닝을 배우고 있습니다."

ex2 = ngram.diff_ngram(a, b, 2)
print("2gram 유사도 :", ex2)

ex3 = ngram.diff_ngram(a, b, 3)
print("3gram 유사도 :", ex3)
```

```python
a = "본문과 전혀 관계 없는 내용이지만 마시멜로는 맛있습니다."
b = "마시멜로는 본문과 전혀 관계 없이 맛있습니다."

ex2 = ngram.diff_ngram(a, b, 2)
print("2gram 유사도 :", ex2)

ex3 = ngram.diff_ngram(a, b, 3)
print("3gram 유사도 :", ex3)
```

#### 코사인(cosine) 유사도
: 두 문장을 비교하여 유사도 측정 

```python
a = "머신러닝은 매우 재미있는 기술이라 공부하고 있습니다."
b = "공부하면 재미있는 기술이라 머신러닝을 배우고 있습니다."

from sklearn.feature_extraction.text import TfidfVectorizer
sentences = (a, b)

tfid_vectorizer = TfidfVectorizer()
sentences
```

```python
# 문장 벡터화(숫자 만들기) 하기 (어휘사전 만들기)
```

```python
# 문장 벡터화 하기 (어휘 사전 만들기)
tfid_matrix = tfid_vectorizer.fit_transform(sentences)
```

```python
### 코사인 유사도 
from sklearn.metrics.pairwise import cosine_similarity

# 첫번째 문장과 두번째 문장 비교 
cos_similar = cosine_similarity(tfid_matrix[0],tfid_matrix[1])
print('코사인 유사도 :',cos_similar) # 코사인 유사도 0.33이면 되게높은거다 
```

#### 유사도를 이용한 추천 시스템 구현
: 코사인 유사도만으로 영화의 줄거리에 기반하여 영화를 추천하는 추천 시스템 구축

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
data = pd.read_csv('../Data/movies_metadata.csv',low_memory=False)
data
```

```python
data.head(2)
```

```python
data.columns
```

> 코사인 유사도에 사용할 데이터는 영화 제목에 해당하는 title, 줄거리에 해당하는 overview
좋아하는 영화를 입력하면 해당 영화의 줄거리와 유사한 줄거리의 영화를 찾아서 추천하는 시스템을 만들 것 입니다.


```python
data[['title', 'overview']].head()
```

```python
data.shape
```

```python
# # 상위 2만개의 샘플
# data = data.head(20000)
```

```python
# overview에 결측치 _ 항상 확인해야 함
data['overview'].isnull().sum()
```

```python
# 결측치를 빈값으로 대체
data['overview'] = data['overview'].fillna('')
```

```python
tfidf = TfidfVectorizer(stop_words='english')
tfid_matrix = tfidf.fit_transform(data['overview'])
tfid_matrix.shape
```

```python
cosine_sim = cosine_similarity(tfid_matrix, tfid_matrix) # 반복문이 일어난다_ 같은거 두개 집어 넣음
print("코사인 유사도 연관 결과 :", cosine_sim.shape)
```

```python
# 기존 데이터프레임으로 부터 영화의 타이틀을 key 영화의 인덱스를 value로 하는 딕셔너리 
title_to_index = dict(zip(data['title'],data.index))
idx = title_to_index['Father of the Bride Part II']
print(idx)
# 영화의 제목을 추천할거라 타이틀이 키가 된다
```

```python
# 선택한 영화의 제목을 입력하면 코사인 유사도를 통해 overview가 유사한 10개의 영화를 찾아내는 함수

def get_recommendations(title, cosine_sim=cosine_sim):
    # 선택한 영화의 타이틀로 부터 해당 영화의 인덱스를 받아 온다.
    idx = title_to_index[title]

    # 해당 영화와 모든 영화와의 유사도를 가져온다
    sim_scores = list(enumerate(cosine_sim[idx]))

    # 유사도에 따라 영화들을 정렬한다.
    sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)

    # 가장 유사한 10개의 영화를 받아온다.
    sim_scores = sim_scores[1:11]

    # 가장 유사한 10개의 영화의 인덱스를 얻는다.
    movie_indices = [idx[0] for idx in sim_scores]

    # 가장 유사한 10개의 영화의 제목을 리턴한다.
    return data['title'].iloc[movie_indices]
```

```python
# 영화 다크 나이트 라이즈의 overview가 유사한 영화
get_recommendations('The Dark Knight Rises')
get_recommendations('Toy Story')
```
