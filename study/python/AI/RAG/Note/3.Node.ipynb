### 1. Node 객체 직접 생성

```python
from llama_index.core import Document
from llama_index.core.schema import TextNode


document = Document(
    text = """인공지능은 우리의 미래를 변화시킬 것 입니다.
                이러한 변화에 우리는 준비되어 있어야 합니다.

"""
)

# 문서를 2개의 노드로 분할 
node1 = TextNode(text = document.text[:24],doc_id =document.id_)
node2 = TextNode(text = document.text[25:],doc_id =document.id_)

print(node1)
print(node2)
```

### 2. 문서를 불러와서 Document와 Node 객체 생성

```python
from llama_index.core import SimpleDirectoryReader
from llama_index.core.node_parser import SimpleNodeParser
```

```python
%pip install docx2txt
```

```python
documents = SimpleDirectoryReader(
    input_dir = "../Data",
    required_exts = [".docx"]
).load_data()
```

```python
documents[0].text
```

```python
parser = SimpleNodeParser.from_defaults(
    chunk_size=100,# 100자씩 분할
    chunk_overlap=20,# overlap 20자_overlap 쓰는 이유 : 분할하면 어느 부분인지 몰라서 연결하는 위치를 알려주기 위해서 씀
)

nodes = parser.get_nodes_from_documents([documents[0]])
```

```python
nodes[0].text
```

```python
# Node 확인
for i, node in enumerate(nodes):
    print(f"=== Node {i+1} ===")
    print(f"Text: {node.text}")
```
