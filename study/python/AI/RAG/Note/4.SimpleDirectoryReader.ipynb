```python
from llama_index.core import SimpleDirectoryReader

documents = SimpleDirectoryReader("../Data").load_data()
```

```python
print("디렉토리 파일 갯수",len(documents))
for document in documents:
    print(document.metadata['file_name'])
```

```python
# 특정 파일만 로드
documents = SimpleDirectoryReader(
    input_dir="../Data",
    input_files=['../Data/json_example.json','../Data/txt_example.txt']
).load_data()

print(len(documents))
print(documents[0].text)
```

```python
# 특정 확장자만 로드
documents = SimpleDirectoryReader(
    input_dir="../Data",
    required_exts=['.csv']
).load_data()

print(len(documents))
print(documents[0].text)
```

```python
# 이미지만 로드
documents = SimpleDirectoryReader(
            input_dir="../Data",
            required_exts=['.jpg', '.jpeg', '.png']
).load_data()

print(len(documents))
# 텍스트는 없어서 지움

from PIL import Image
image_path = documents[0].metadata['file_path']
image_path
# 그림 확인
Image.open(image_path)
```
