#### 기초 통계량 산출

```r
cafe <- read.csv("../Data/coffee.csv")
head(cafe)
```

```r
# 빈도수
table(cafe$Coffee)
```

```r
# 범위 빈도수
coffee_range = seq(0, 60, 10)
table(cut(cafe$Coffee, breaks = coffee_range, right = F))
```

```r
# 줄기-잎 그림
stem(cafe$Coffee) # 갯수 파악
```

```r
# Coffee와 쥬스의 변동계수 구하기 
(cafe$Coffee)
```

```r
# 평균, 표준편차, 변동계수 계산
coffee.mean <- mean(cafe$Coffee)
coffee.sd <- sd(cafe$Coffee)
coffee.cv <- coffee.sd / coffee.mean

juice.mean <- mean(cafe$Fruit.Cup)
juice.sd <- sd(cafe$Fruit.Cup)
juice.cv <- juice.sd / juice.mean

data.frame(
  음료 = c("Coffee", "Juice"),
  평균 = c(coffee.mean, juice.mean),
  표준편차 = c(coffee.sd, juice.sd),
  변동계수 = c(coffee.cv, juice.cv)
)
```
