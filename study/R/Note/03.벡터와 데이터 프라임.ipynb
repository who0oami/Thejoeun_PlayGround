#### 문자열 처리
수집한 데이터의 필요한 문자열을 자르고, 교체하고 수출하는 작업

## stringr package 사용 ( 파이썬엔 없음 )
- str_length() : 문자열 길이
- str_c() : 문자열 연결
- str_sub() : 범위에 해당하는 문자열 생성
- str_split() : 구분자를 기준으로 문자열을 분리하여 부분 생성
- str_replace() : 기존 문자열을 특정 문자열로 변환
- str_extract() : 문자열에서 특정 문자열 패턴의 첫번째 문자열 추가
- str_extract_all() : -> 모든 문자열 추출
- str_locate() : 문자열에서 특정 문자열 패턴의 첫번째 위치 추출
- str_locate_all() : -> 모든 문자열 추출 

```r
# install.packages("stringr")  # pc에 한번 설치하고 불러올땐 라이브러리로 
```

```r
library(stringr) # 문자열 처리 할 게 있으면 무조건 이거씀_ 편해서
```

```r
string <- "hongkildong105youbi1002kwanwoo25장비2005"
```

```r
# a ~ z 까지의 영문자가 연속으로 3개 발견된 문자열 추출
str_extract_all(string, "[a-z]{3}") # 뒤에 패턴을 준것
```

```r
# a ~ z 까지의 영문자가 연속으로 3개~5개 발견된 문자열 추출
str_extract_all(string, "[a-z]{3,5}") # 뒤에 패턴을 준것
```

```r
# a~z까지의 영문자가 연속으로 1개 이상 발견된 문자열 추출
str_extract_all(string, "[a-z]{1,}")
```

```r
# a~z까지의 영문자가 연속으로 1개이상 발견된 문자열 추출 
aaa <- str_extract_all(string,"[a-z]{1,}")
class(aaa)
bbb <- unlist(aaa)
bbb
```

```r
# 숫자만 추출하기
str_extract_all(string,("[a-z]{1,}"))
```

```r
str_extract_all(string, "[0-9]{1,}")
```

```r
as.numeric(unlist(str_extract_all(string, "[0-9]{1,}")))
```

```r
# 한글만 추출하기
str_extract_all(string,("[ㄱ-ㅎ]{1,}"))
```

```r
str_extract_all(string, "[가-히]{1,}")
```

```r
str_extract_all(string, "[각-힣]{1,}")
```

```r
str_extract_all(string, "[가-힣]{1,}")
```

```r
# 영문자를 제외한 문자 출력하기 _ 정규식에쓰임
str_extract_all(string,"[^a-z]{1,}")
```

```r
# 문자열 길이 구하기
str_locate(string,"장비") # 장비의 앞에 있는 거 가져옴
```

```r
str_sub(string,1,32)
str_sub(string,1,str_locate(string,"장비")[1]-1) # 장비 위치가 바뀌어도 잘 가지고 오니까 이렇게 씀
```

```r
# 대문자 소문자 변경하기
str_to_upper(string)
str_to_lower(string)
```

```r
# 문자열 교체하기
string <- str_replace(string, "hongkildong105", "홍길동35,")
string <- str_replace(string, "yoube1002", "유비45,")
string <- str_replace(string, "kwanwoo25", "관우25,")
string <- str_replace(string, "2005", "20")
string
```

```r
# 문자열 분리하기
str_split(string,",")
```

```r
# 문자열 분리하기
unlist(str_split(string,","))
```

```r
# 천단위 숫자가 들어간 문자열을 숫자로 변경하기
price <- "123,456,789"
as.numeric(str_replace_all(price,",",""))
as.numeric(str_remove_all(price,","))
```

```r
# Na를 변경하기
temp <- c(1, 2, NA, 4)
temp
str_replace_all(str_replace_na(temp), 'NA', "3")
```
