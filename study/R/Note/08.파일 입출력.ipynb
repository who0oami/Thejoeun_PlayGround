#### 서울 교통사고 조사 데이터셋

```r
ar(seoul$'발생건수')
```

```r
# 자치구명별 발생건수의 평균
regionAcc.mean <- aggregate(발생건수~자치구명,seoul, mean)
head(regionAcc.mean)
```

```r
# 내림차순으로 정렬
arrange(regionAcc.mean, desc(발생건수))
```

```r
# 자치구별 표준편차
regionAcc.sd <- aggregate(발생건수~ 자치구명,seoul,sd)
regionAcc.sd
```

```r
regionAcc.mean
```

```r
(regionAcc.sd$발생건수) / (regionAcc.mean$발생건수)
```

```r
# 자치구별 발생건수의 변동계수
# 변동계수 = 표준편차 / 평균
regionAcc.cv <- (regionAcc.sd$발생건수) / (regionAcc.mean$발생건수)
regionAcc.cv
```

```r
regionAcc.cv <- aggregate(발생건수 ~ 자치구명, seoul, ((regionAcc.sd$발생건수) /(regionAcc.mean$발생건수)))
regionAcc.cv
```

```r
# 자치구명과 변동계수를 가진 DataFream 만들기기
regionAcc <- data_frame(자치구명=regionAcc.mean$'자치구명',변동계수=regionAcc.cv)
regionAcc
```

```r
# 교통사고 발생건수의 변동계수로 Top10 출력
head(arrange(regionAcc,변동계수),10)
```
