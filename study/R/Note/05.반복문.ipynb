```r
# excel에서 q3 컬럼 데이터의 홀수의 합과 짝수의 합 구하기 
# 출력 홀수의 합은 _ 이고 짝수의 합은 _입니다
excel <- read.csv("../Data/excel.csv")
head(excel)
num1 <- 0
num2 <- 0
for (i in excel$q3){
    if(i%%2 == 0){
        num1 <- num1 + i
    }else{
    num2 <- num2 + i
    }
}
cat("홀수의 합은",num2,"이고","짝수의 합은",num1,"입니다")
```

#### 함수

```r
gugu <- function(dan){
    for(i in 1:9){
        cat(dan, "X", i, "=",dan*i, "\n")
    }
}
```

```r
gugu(5)
```

```r
seq(1,10)
```

```r
# gugu()를 이용한 구구단 전체 출력
gugu(seq(2,9))
```

```r
for (i in seq(2,9)){
    cat(i,"단","======","\n")
    cat("\n",gugu(i))
}
```

```r
# gugu()를 이용한 구구단 전체 출력 _ 선생님 정답
```

```r
for(dan in 2:9){
    cat(">>>", dan, "단", "<<<\n")
    gugu(dan)
    cat("-----------------\n")
}
```

```r
gugu <- function(dan, num){
    for(i in 1:num){
        cat(dan, "X", i, "=",dan*i, "\n")
    }
}
```

```r
gugu(2, 12)
```

```r
# 2개의 벡터 자료를 함께 출력하기
score <- c(85, 95, 89)
name <- c("유비", "관우", "장비")
```

```r
for (num in score){
    for(n in name){
        print(n, num)
    }
}
```

```r
for(i in 1:length(score)){
    cat(name[i], score[i], "\n")
}
```

```r
# 선생님 정답_ paste를 활용해야 함

cat(paste(name, ";", score, "\n"), sep="")
```

#### while

```r
i <- 0 
while(i<10){
    i<-i+1
    print(i)
}
```

#### repeat

```r
i <- 0
repeat{
    i <- i+1
    if(i>10){
        break
    }
    print(i)
}

# repeat 는 소괄호가 없음
```
