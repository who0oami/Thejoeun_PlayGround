```r
set.seed(1234) # 랜덤 시드 똑같이 맞추려고
library(caret)
samp <- createDataPartition(iris$Species, p = 0.7, list = F) # 70%의 데이터를 랜덤하게 뽑아서 인덱스 번호로 반환
```

```r
# 학습용
data.tr <- iris[samp,]

# 검증용
data.test <- iris[-samp,] # samp뺴고 나머지
```

```r
nrow(data.tr)
nrow(data.test)
```

```r
# 문제 및 정답 만들기
x <- subset(data.test, select = -Species) # 문제
y <- data.test$Species # 정답
```

```r
install.packages("randomForest")
```

```r
library(randomForest)
```

```r
# 학습시키기
rf <- randomForest(Species ~ ., data = data.tr, ntree = 100)
```

```r
# 검증
pred <- predict(rf, x)
```

```r
pred
```

```r
# 예측력
mean(pred==y)
```

```r
# 오분류표
table(pred, y)
```
