```r
# 2행 2열 차트
par(bg = 'white')
price <- runif(10, 1, 100) # 데이터, min, max로 쓰면 파라메터 안써도 되고, 순서 바꾸면 써야 함

par(mfrow = c(2,2))
plot(price, type = 'l')
plot(price, type = 'o')
plot(price, type = 'h')
plot(price, type = 's')
```

```r
# 2행 2열 차트
par(bg = 'white')
price <- runif(10, 1, 100) 

par(mfrow = c(2,2))
plot(price, type = 'o', pch = 5)
plot(price, type = 'o', pch = 15)
plot(price, type = 'o', pch = 20, col = 'blue')
plot(price, type = 'o', pch = c(rep(15, 8), 20, 20),
                            col='orange',
                            cex=c(rep(1,8),5,1) # 크기조절
)
```

```r
# 낙서하기

par(family = 'AppleGothic', bg = 'white')

x <- c(1, 3, 6, 8, 9)
y <- c(12, 56, 78, 32, 9)
plot(x,y)

# (3,3) 과 (4,4)를 연결하는 선분 그리기
segments(6, 78, 8, 32)

# 화살표 그리기
arrows(3, 56, 1, 12, col = 'blue', lwd = 3)

# 사각표 그리기
rect(4, 20, 6, 30, col='red', density = 30) # density는 사각형 내부의 선의 간격_ 숫자가 클수록 간격이 좁아짐 

# 글자 쓰기
text(4, 40, "This is a  샘플", srt=50) # srt는 글자 회전 각도

# 테두리 활용하기
mtext("character", side = 4, adj=0.5) # side는 1,2,3,4로 아래, 왼쪽, 위, 오른쪽을 의미한다. adj는 글자 위치 조절

# x축의 추가 (y축의 40 위치에 0~10까지 표시하고 색은 빨간색으로)
axis(1, pos=40, at=0:10, col='red') # 파이썬에 없음

# y축의 추가 (x축의 6 위치에 20~60까지 표시하고 색은 파란색으로)
axis(2, pos=6, at=20:60, col='blue') # seq는 20부터 60까지 숫자 생성
```

#### 수학 함수의 시각화

```r
f <- function(x) {
    return(x^2)
}
```

```r
x <- 1:5
f(x)
```

```r
par(bg = 'white')
x <- seq(-5, 5, 0.1)

plot(x, f(x), type = 'l') 
# y = x²
```

```r
# y = x² + 2x + 3
plot(x, f(x), type = 'l')
```
