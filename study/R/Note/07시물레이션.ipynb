### 몬테카를로 시뮬레이션

```r
example(runif)
```

```r
coin <- function(num){
    count1 <- 0
    count2 <- 0
    for(i in 1:num){
    if(runif(1) <= 1/2){
        count1 <- count1 + 1
    }else{
        count2 <- count2 + 1
    }
    }
cat("앞면",count1 ,"\n","뒷면",count2)
}
```

```r
table(coin(10))
table(coin(100))
table(coin(1000))
table(coin(10000))
table(coin(100000))
table(coin(1000000))
```

```r
dice <- function(n){
    diceFace <- rep(0,6)
    cat("Throw:",n,"\n")
    for(i in 1:n){
        x <- trunc(runif(1) * 6) + 1
        diceFace[x] <- diceFace[x] +1
    }

    for(i in 1:length(diceFace)){
        cat("No", i ,"=>",diceFace[i],"\n")
    }
    cat("-----------------\n")

}
```

```r
table(dice(10))
table(dice(100))
table(dice(1000))
table(dice(10000))
table(dice(100000))
table(dice(1000000))
```
