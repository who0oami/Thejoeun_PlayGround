```r
# plot type별 시각화
par(mfrow = c(3, 3), bg = 'white', mar = c(3, 3, 2, 1))
y <- c(1, 3, 2, 4, 8, 5, 7, 6, 9)
types <- c('p', 'l', 'b', 'c', 'o', 'h', 's', 'S', 'n')
for (t in types) {
    plot(
        y,
        type = t,
        xlim = c(min(y) - 1, max(y) + 1),
        ylim = c(min(y) - 1, max(y) + 1),
        main = paste0('type = ', t)
    )
}
```

```r
# 꺽은선 그래프 

par(bg='white')
y <- c(1,3,2,4,8,5,7,6,9)
plot(
    y,
    xlim=(c(min(y)-1,max(y+1))),
    ylim=(c(min(y)-1,max(y+1))),
    main = "Simple Time Series",
    type = 'b',
    lty = 6, # 0~6 까지 종류
    pch = ".",
    col = 'red'
    )
```

```r
# plot에서 pch 기호 종류별 시각화
par(mfrow = c(2, 5), bg = 'white', mar = c(3, 3, 2, 1))
y <- c(1, 3, 2, 4, 8, 5, 7, 6, 9)
pch_values <- 0:25
for (pch in pch_values) {
    plot(
        y,
        pch = pch,
        cex = 2,
        xlim = c(min(y) - 1, max(y) + 1),
        ylim = c(min(y) - 1, max(y) + 1),
        main = paste0('pch = ', pch)
    )
}
```

```r
# 3종류 data의 선그래프
abc <- c(260,300,250,280,310)
def <- c(180,200,210,190,170)
ghi <- c(210,250,260,210,270)
```

```r
par(family = 'AppleGothic',bg = 'white')
plot(abc,
    type ='o',
    col = 'red',
    ylim = c(0,max(abc)+150),
    axes = F, #좌표 Frame 제거
    ann = F # label 제품
    )


    # x축과 x축의 라벨 설정
    axis(1,at = 1:5,lab = c('서울','대전','대구','광주',"원주"))

    #y축과 y축의 라벨을 설정
    axis(2,ylim = c(0, max(abc)+100))

    #main title 
    title(
        main = '과일 판매량',
        col.main ='red',
        cex.main = 2
    )

    # x축 label
    title(
        xlab = "지역",
        col.lab = 'blue',
        cex.lab = 2
    )
    
    # y축 label
    title(
        ylab = '판매량',
        col.lab ='blue'
    )
    
    # 선그리기로 def 와 ghi 그리기
    lines(
        def,
        type = 'o',
        pch = 22,
        col = 'green',
        lty =2,
    )

    lines(
        ghi,
        type = 'o',
        pch = 22,
        col = 'blue',
        lty =2,
    )


    #범례
    legend(4,350,
        c("야구장", "축구장","해변가"),
        col = c('red', 'green', 'blue'),
        pch = c(21, 22, 22),
        lty = c(1, 2, 2),
        cex = 0.8
    )
```

```r
par(family = 'AppleGothic',bg = 'white')
Sales <- c(210,110,400,550,600,130)
week <- c("월","화","수","목","금","토")
ratio <- Sales / sum(Sales) * 100
label <- paste(week,"\n",ratio,"%")


pie(Sales,
    main = "영어 실적 파이차트",
    col = rainbow(length(Sales)),
    labels = label
)
```
