#### Vector 자료구조
: 벡터는 R에서 가장 기본이 되는 자료구조 이다.
- 1차원 선형 자료 구조 형태.
- 구성 : 변수[index]형태로 구성, index는 1부터 시작
- 동일한 자료형의 데이터만 저장할 수 있다.
- 벡터 처리 함수 : sum(), mean(), sd(),union(), ...... (모든 언어가 같게 되어있음)

#### 벡터 자료 처리

```r
x <- c(1, 3, 5, 7)
y <- c(3, 5)

# 합집합
union(x, y)

# 교집합
intersect(x, y)

# 차집합
setdiff(x, y)
```

```r
# 벡터 자료 정렬
sample(1:10, 2)

# 자료 섞기 _ suffle
sample(1:10, 10)

# 정렬
sort(sample(1:10, 10)) # 오름차순
sort(sample(1:10, 10), decreasing =  T) # 내림차순
```
