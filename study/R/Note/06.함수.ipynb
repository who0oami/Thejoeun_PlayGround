#### 기술통계량을 계산하는 함수 만들기

```r
test <- read.csv("../Data/test.csv")
head(test)
```

```r
# 데이터 갯수
nrow(test)
```

```r
# 구성
str(test)
```

```r
# 기본 통계량
summary(test)
```

```r
# A 컬럼의 빈도수
table(test$A)
```

```r
# 각 컬럼의 빈도수 구하기
for(i in 1:ncol(test)){
    cat(i, "번째 컬럼의 빈도 분석 결과")
    print(table(test[,i]))
    cat("-----------------------\n")
}
```

```r
# 각 컬럼의 빈도수의 최대값, 최소값 구하기
for(i in 1:ncol(test)){
    cat(i, "번째 컬럼의 빈도수의 최대값과 최소값\n")
    f <- table(test[,i])
    cat("최대값:", max(f), ", 최소값:", min(f), "\n")
    cat("-----------------------\n")
}
```

```r
# 빈도 분석결과와 최대값 최소값 출력하는 함수 만들기
tableMaxMin <- 

tableMaxMin <- function(test){
    for(i in 1:ncol(test)){
    cat(i, "번째 컬럼의 빈도수의 최대값과 최소값\n")
    f <- table(test[,i])
    cat("최대값:", max(f), ", 최소값:", min(f), "\n")
    cat("-----------------------\n")
}
}
```

```r
tableMaxMin(test)
```

```r
gugu <- function(dans, nums){
    for(i in dans){
        for(num in nums){
            cat(i, "X", num, "=",i*num, "\n")
        }
    }
}
```

```r
gugu(2:3, 1:5)
```

```r
# 정해진 범위의 구구단 출력 함수 _ 선생님
gugu <- function(dans, nums){
    for(dan in dans){
        for (num in nums){
            cat(dan, "X", num, "=",dan*num, "\n")
        }
    }
}
```

```r
gugu(2:3, 1:5)
```

```r
x <- c(2:3)
y <- c(1:5)
gugu(x, y)
# 위에랑 같은거
```

#### 결측치(NA) 자료 평균 계산

```r
data <- c(10, 20, 5, 4, 40, 7, NA, 6, 3, NA, 2, NA)
```

```r
mean(data)
```

```r
# NA의 갯수 구하기
sum(is.na(data)) # na 를 확인하는 방법은 is.na 밖에 없다
table(is.na(data)) # 빈도수
```

```r
# 결측치를 무조건 제거하면 통계량을 얻을 수 없으며, 데이터의 손실로 문제 발생
# 보통 회귀함수(흐름을 파악)로 데이터를 파악한 뒤 처리함
mean(data, na.rm = T)
```

```r
# 1) NA를 0으로 대체
data_0 <- ifelse(is.na(data), 0, data) # data가 있으면 넣어주고 없으면 0으로 바꿔줌
mean(data_0)
```

```r
# 2) NA를 평균으로 대체 => 아무의미 없음, 이렇게 하면 안된다 _ 평균은 계산값이지, 데이터가 아님
data_mean <- ifelse(is.na(data), mean(data, na.rm = T), data) 
mean(data_mean)
```

```r
# 3) NA를 중앙값으로 대체 _ 중앙값은 데이터임
data_median <- ifelse(is.na(data), median(data, na.rm = T), data) 
mean(data_median)
```

```r
# 3) NA를 최대값으로 대체
data_max <- ifelse(is.na(data), max(data, na.rm = T), data) 
mean(data_max)
```

```r
# 4) NA를 최소값으로 대체
data_min <- ifelse(is.na(data), min(data, na.rm = T), data) 
mean(data_min)
```
